```python
# 딕셔너리에서 Key를 사용해 Value 얻기
grade = {'pey': 10, 'julliet' : 99}
print(grade['pey'])
print(grade['julliet'])
```

```python
# Key는 고유한 값이므로 중복되는 Key 설정시
# 하나를 제외한 나머지 모든 중복 Key는 무시된다
# Key값에 리스트는 넣을 수 없다
a={1:'a',1:'b'}
a
```

```python
# 딕셔너리 관련 함수
# (1) Key 리스트 만들기 keys
a = {'name':'pey','phone':'010-9791-9256','birth':'0901'}
print(a.keys())
# dict_keys를 리스트로 반환하기
print(list(a.keys()))
# for print()는 들여쓰기 필수
for k in a.keys():
    print(k)
```

```python
# (2) Value 리스트 만들기 values
a = {'name':'pey','phone':'010-9791-9256','birth':'0901'}
print(a.values())
# (3) Key,Values 쌍 얻기 items
print(a.items())
# (4) Key:Value 모두 지우기 clear
a.clear()
print(a)
# (5) Key로 Value 얻기 get
a = {'name':'pey','phone':'010-9791-9256','birth':'0901'}
print(a.get('name'))
print(a.get('phone'))
```

```python
# (6) 딕셔너리에 존재하지 않는 키
a = {'name':'pey','phone':'010-9791-9256','birth':'0901'}
# a['nokey'] 는 오류를 발생시킴
# a.get('nokey')는 None을 반환함
print(a.get('nokey'))
```

```python
# (7) 딕셔너리안에 key가 존재하는지 조사하기 in
a = {'name':'pey','phone':'010-9791-9256','birth':'0901'}
print('name' in a)
print('email' in a)
```

```python
# (8) Key로 Value 얻기 pop
# pop 은 해당항목을 삭제함]
a = {'name':'pey','phone':'010-9791-9256','birth':'0901'}
phone = a.pop('phone')
print(phone)
print(a)
# 존재하지 않는 Key에 대해서 디폴트 값 지정 가능
# JAVA의 Try-Catch와 비슷 Error를 막고 안전한 값을 주는 방식
email = a.pop('email','정보없음')
print(email)
print(a)
```

```python
# 중괄호{}를 사용해서 집합 자료형을 직접 생성 가능
s3 = {1,2,3}
print(s3)
s4= {'a','b','c'}
print(s4)
```

```python
# 집합 자료형에 저장된 값을 인덱싱으로 접근하는 방법
# 리스트나 튜플로 변환 후 접근해야 한다
#=======리스트=======
s1 = set([1,2,3])
l1 = list(s1)
print(l1)
print(l1[0])
#=======튜플=======
s = set([1,2,3])
t = tuple(s)
print(t)
print(t[0])
```

```python
# 교집합 구하기
sa = set([1,2,3,4,5,6])
sb = set([4,5,6,7,8,9])
print(sa&sb)
print(sa.intersection(sb))
```

```python
# 합집합 구하기
sa = set([1,2,3,4,5,6])
sb = set([4,5,6,7,8,9])
print(sa | sb)
print(sa.union(sb))
```

```python
# 차집합 구하기
sa = set([1,2,3,4,5,6])
sb = set([4,5,6,7,8,9])
print(sa-sb)
print(sb-sa)
print(sa.difference(sb))
print(sb.difference(sa))
```

```python
# 집합 자료형 관련 함수
# (1) 값 1개 추가하기 add
s1 = set([1,2,3])
s1.add(4)
print(s1)
```

```python
# (2) 값 여러개 추가하기 update
s1 = set([1,2,3])
s1.update([4,5,6])
s1
```

```python
# (3) 특정 값 제거하기 remove
s1 = set([1,2,3])
s1.remove(2)
print(s1)
# (4) 특정 값 제거하기 discard
s2 = set([1,2,3])
s2.discard(2)
print(s2)
print(s2.discard(4)) # 없는값도 오류 발생 안함
print(s2)
```

```python
# (5) 모든 값 제거하기 clear
s1 = set([1,2,3])
s1.clear()
s1
```

```python
# 자료형의 참과 거짓
# 문자열, 리스트, 튜플, 딕셔너리 등의 값이 비어있으면 거짓이 되고,
# 비어있지 않으면 참이 된다.
# 숫자에서는 그 값이 0이 될 때 거짓이 된다.
a = [1,2,3,4]
while a:
    print(a.pop())
print(a)
# 마지막 요소를 계속 끄집어내서 삭제 처리하고 삭제할게 더 이상 없으면
# 빈리스트가 되어 거짓이 됨. while문을 빠져나가게 돼서 종료
```

```python
# 요소값이 존재하므로 "참"을 반환함
if [1,2,3]:
  print("참")
else:
  print("거짓")
```

```python
# 논리연산자
# (1) and : 양쪽 조건이 모두 참 일때만 True 반환
print(True and True)
print(True and False)
print(False and True)
print (False and False)
```

```python
# id는 변수가 가리키고 있는 객체의 주소 값을 반환하는
# 파이썬의 내장함수
a = [1,2,3]
id(a)
```

```python
# 리스트를 복사할 때
a = [1,2,3]
b = a
print(id(a))
print(id(b))
# a 와 b가 같은 객체를 가리키고 있는지
print (a is b)
#===============================
a[1] = 4
print(a)
print(b)
```

```python
# [:] 이용하기
# b = a -> a와 b가 같은 리스트 객체를 가리킴 (참조 개념)
# b = a[:] -> a와 동일한 내용을 가진 새로운 리스트를 생성함
#             a의 요소값이 바뀌더라도 b의 요소값은 바뀌지않음
a = [1,2,3]
b = a[:]
a[1] = 4
print(a)
print(b)
```

```python
# copy  모듈 이용하기
# [:] 와 같은 개념이므로 b is a 는 False
from copy import copy
a = [1,2,3]
b = copy(a)
print (b is a)
```
