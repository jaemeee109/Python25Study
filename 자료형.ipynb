```python
# 사칙연산 ( + - * / )
a = 3
b = 4
print(a + b)
print(a - b)
print(a*b)
print(a/b)
```

```python
# x의 y제곱을 나타내는 ** 연산자
a = 3
b = 4
print(a ** b)
```

```python
# 나눗셈 후 나머지를 반환하는 % 연산자
print(7 % 3)
print(3 % 7)
```

```python
# 나눗셈 후 몫을 반환하는 // 연산자 (정수값만 반환)
print (7/4)
print (7//4)
```

```python
# 복합 연산자
# +=, -=, /=, //=, %=, **=

a = 1 # a = 1
a = a + 1
print (a) # 1+1로 인해 2가 된 a를 반환
#===============================
b = 1
b += 1 # b = b + 1과 동일한 연산식
print (b)
#===============================
c = 1
c = c - 1
print (c)
#===============================
d = 1
d -= 1 # c = c -1 과 동일한 연산식
print(d)
```

```python
# 문자열 인덱싱
a = "Life is too short, You need Python"
print(len(a))
print(a[3])
print(a[6])
print(a[33])
# 0부터 계산이기 때문에 a[34]를 넣으면 오류발생
print(a[-3])
print(a[-34])
# -는 뒤에서부터 해당 자리에 있는 문자를 반환
```

```python
# 문자열 슬라이싱
#start 인덱스 이상 end인덱스 미만이라 [4]는 포함되지 않음
a = "Life is too short, You need Python"
b = a[0] + a[1] + a[2] + a[3]
print(b)
print(a[0:4])
#end를 생략하면 start부터 끝까지를 출력
print(a[10:])
#start를 생략하면 처음부터 end까지 출력
print(a[:8])
#start와 end 모두 생략하면 처음부터 끝까지 출력
print(a[:])
# - 기호 사용 가능
print(a[19:-7])
```

```python
# 슬라이싱으로 문자열 나누기
a = "20251030thursday"
date = a[:8]
week1 = a[8:]
year = '2025'
day = '1030'
week2 = 'thursday'
print(date)
print(week1)
print(year)
print(day)
print(week2)
```

```python
# format 함수를 사용한 포매팅
# 숫자 바로 대입
print("I eat {0} apples".format(3))
# 문자열 바로 대입
print("I eat {0} apples".format("five"))
# 숫자 값을 가진 변수로 대입하기
number=3
print("I eat {0} apples".format(number))
# 2개 이상의 값 넣기
num = 10
day = "three"
print("I eat {0} apples. so I was sick for {1} days.".format(num, day))
# 이름으로 넣기
print("I ate {num} apples. so I was sick for {day} days.".format(num = 10, day =3))
# 인덱스와 이름을 혼용해서 넣기
print("I ate {0} apples. so I was sick for {day} days.".format(10, day=3))
```

```python
# 왼쪽정렬 :<
"{0:<10}".format("hi")
```

```python
# 오른쪽 정렬 :>
"{0:>10}".format("hi")
```

```python
# 가운데 정렬 :^
"{0:^10}".format("hi")
```

```python
# 공백 채우기
print("{0:=^10}".format("hi"))
print("{0:!^10}".format("hello"))
print("{0:*^10}".format("bye"))
```

```python
#  소수점 표현하기
y = 3.42134234
print("{0:0.4f}".format(y))
print("{0:10.4f}".format(y))
```

```python
# f 문자열 포매팅
name = "홍길동"
age = 30
print(f'나의 이름은 {name} 입니다. 나이는 {age} 입니다')
print(f'나는 내년이면 {age+1}살이 된다')
#=============================
d = {name, age}
print(f"나의 이름은 {d[name]} 입니다. ")
```

```python
print("========정렬=========")
# 왼쪽 정렬
print(f'{"hi":<10}')
# 오른쪽 정렬
print(f'{"hi":>10}')
# 왼쪽 정렬
print(f'{"hi":^10}')
print("========공백 채우기=========")
# 가운데 정렬
print(f'{"hi":*^10}')
# 왼쪽 정렬
print(f'{"hi":!<10}')
print("========소수점=========")
# 4자리까지만 표현
y = 3.42134234
print(f'{y:0.4f}')
# 4자리까지 표현하고 자리수 10으로 맞춤
print(f'{y:10.4f}')
```

```python
# 문자열이 알파벳으로만 구성되어 있는지 확인 isalpha
a = "Python"
print(a.isalpha())
# 숫자가 포함 될 경우
b = "Python3"
print(b.isalpha())
# 공백이 포함 될 경우
c = "Hello World"
print(c.isalpha())
```

```python
# 문자열이 특정 문자(열)로 끝나는지 확인 endswith
s = "Life is too short"
print(s.endswith("short"))
print(s.endswith("too"))
```
