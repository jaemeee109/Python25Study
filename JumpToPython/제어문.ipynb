```python
# money에 True를 대입했으므로 money는 참이다
# 따라서 '택시를 타고 가라" 가 출력된다
money = True
if money:
    print("택시를 타고 가라")
else:
    print("걸어 가라")
```

```python
# 비교 연산자
# x > y : x가 y보다 크다
# x < y : x가 y보다 작다
# x == y : x와 y 가 같다
# x != y : x와 y가 같지 않다
# >= y : x가 y보다 크거나 같다
# x <= y : x가 y보다 작거나 같다
x = 3
y = 2
print(x > y)
print(x < y)
print(x == y)
print(x != y)
```

```python
# 만약 3,000원 이상의 돈을 가지고 있으면 택시를 타고 가고,
# 그렇지 않으면 걸어가라
money = 2000
if money >= 3000 :
  print("택시를 타고 가라")
else:
  print("걸어가라")
```

```python
# 조건을 판단하기 위해 사용하는 다른 연산자 and , or, not
# x or y : x와 y 둘 중 하나만 참이어도 참이다
# x and y : x와 y 모두 참이어야 참이다
# not x : x가 거짓이면 참이다
# =======================================================
# 만약 3,000원 이상의 돈을 가지고 있으면 택시를 타고 가고,
# 그렇지 않으면 걸어가라
money = 2000
card = True
if money >= 3000 or card:
  print("택시를 타고 가라")
else:
   print("걸어가라")
# money는 2,000원 이지만, card가 True 이기 때문에 조건문이 참이 됨
```

```python
# in, not in
# x in 리스트 : x not in 리스트
# x in 튜플 : x not in 튜플
# x in 문자열 : x not in 문자열
#====================================
# (1) 리스트
# 1이 [1,2,3] 안에 있으므로 참
print(1 in [1,2,3])
# 1이 [1,2,3] 안에 없냐는 조건인데 있으므로 거짓
print(1 not in [1,2,3])
#====================================
# (2) 튜플
print('a' in ('a','b','c'))
print('j' not in 'python')
#====================================
# (3) 문자열
# 만약 주머니에 돈이 있으면 택시를 타고 가고, 없으면 걸어가라.
pocket = ['paper', 'cellphone','money']
if 'money' in pocket:
  print("택시를 타고 가라")
else:
  print("걸어가라")
```

```python
# 조건문에서 아무것도 실행되지 않게 설정할 때 pass
# 주머니에 돈이 있으면 가만히 있고, 주머니에 돈이 없으면 카드를 꺼내라.
pocket = ['paper', 'cellphone','money']
if 'money' in pocket:
  pass
else:
  print("카드를 꺼내라")
# pocket 안에 money 문자열이 있기 때문에 pass가 수행되고
# 결과값이 나오지 않음
```

```python
# 다양한 조건을 판단하는 elif
# ===========================================
# 주머니에 돈이 있으면 택시를 타고 가고,
# 주머니에 돈은 없지만 카드가 있으면 택시를 타고 가고,
# 돈도 없고 카드도 없으면 걸어가라.
# ===========================================
# (1) if와 else만으로 표현
pocket = ['paper', 'cellphone']
card = True
if 'money' in pocket:
  print ("택시를 타고 가라")
else:
  if card :
    print ("택시를 타고 가라")
  else:
    print("걸어가라")
# ===========================================
# (2) elif를 사용해서 표현
if 'money' in pocket:
  print("택시를 타고 가라")
elif card:
  print("택시를 타고 가라")
else:
  print("걸어가라")
```

```python
# if문을 한 줄로 작성하기
pocket = ['paper', 'cellphone','money']
if 'money' in pocket:
  pass
else:
  print("카드를 꺼내라")
# => if문 다음에 수행할 문장이 한 줄이고 else문 다음에
# 수행할 문장도 한 줄 밖에 되지 않음
# ------------------------------------------------------------
# 간략하게 코드를 작성하는 방법
pocket = ['paper', 'cellphone','money']
if 'money' in pocket: pass
else: print("카드를 꺼내라")
```
